# Data Analysis and Visualization Demo
This notebook demonstrates how to load a CSV dataset, display its contents as a pandas DataFrame, and plot a visual chart.

---
## Step 1: Environment Setup & Data Creation
We will import the required libraries and programmatically create a CSV file for this demonstration.

```python
!uv pip install seaborn
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Create a dummy dataset
data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'API_Calls_Millions': [1.2, 1.8, 2.5, 3.1, 4.2, 5.5],
    'Cost_USD': [1200, 1550, 2100, 2400, 3100, 3800]
}

# Save to CSV
df_init = pd.DataFrame(data)
df_init.to_csv('metrics_data.csv', index=False)
print("CSV file 'metrics_data.csv' created successfully!")
```

## Step 2: Load and Display the DataFrame
Now we read the CSV file back into a pandas DataFrame and display it below.

```python
# Load CSV
df = pd.read_csv('metrics_data.csv')

# Display DataFrame
df
```

## Step 3: Plot the Visual Chart
Finally, we utilize matplotlib and seaborn to visualize the trend of API calls and costs over time.

```python
# Set style
sns.set_theme(style="whitegrid")

# Plot
plt.figure(figsize=(9, 5))
plt.plot(df['Month'], df['API_Calls_Millions'], marker='o', color='b', linewidth=2, label='API Calls (M)')
plt.title('Monthly API Usage Trend', fontsize=14, fontweight='bold')
plt.xlabel('Month')
plt.ylabel('Calls in Millions')
plt.legend()
plt.tight_layout()

# Render plot
plt.show()
```
